### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv).

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep='\t')
chipo.head()
```

### Step 4. How many products cost more than $10.00?

```python
chipo['float_price'] = pd.to_numeric(chipo.item_price.str.slice(1)) / chipo.quantity
price_per_product = chipo.groupby('item_name').agg({'float_price' : 'max'})

# переводим цены за товар из строкового типа в числовой и находим цену за товар (делим на quantity)
price_per_product.head()
```

```python
price_per_product[price_per_product.float_price > 10].count()
```
