### Step 9. Which was the most-ordered item?

```python
chipo.groupby("item_name")["quantity"].sum().idxmax() #по количеству заказанных товаров
```

```python
chipo.groupby("item_name")["order_id"].nunique().idxmax() #по количеству заказов, в которых заказали данный товар
```

### Step 10. For the most-ordered item, how many items were ordered?

```python
chipo.groupby("item_name")["quantity"].sum().max() # по колиеству заказанных товаров
```

```python
(chipo.groupby("item_name")
.agg(
    number_of_goods=("quantity", 'sum'),
    number_of_orders=("order_id", "nunique")
).sort_values(by="number_of_orders", ascending=False
).iloc[0, 0]) # по количеству заказов, где участвует товар
```

### Step 13. Turn the item price into a float

#### Step 13.a. Check the item price type

```python
chipo["item_price"].info()
```

#### Step 13.b. Create a lambda function and change the type of item price

```python
sus = chipo["item_price"].map((lambda x : float(x[1:-1])))
sus.head()
```

#### Step 13.c. Check the item price type

```python
chipo["item_price"] = sus
chipo.item_price.dtype
```

### Step 14. How much was the revenue for the period in the dataset?

```python
rev = (chipo["quantity"] * chipo["item_price"]).sum()
print("$ " + str(rev))
```

### Step 15. How many orders were made in the period?

### Step 16. What is the average revenue amount per order?

```python
# Solution 1
rev / orders_num
```
